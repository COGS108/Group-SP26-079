```python
import pandas as pd
import numpy as np
from scipy import stats
from scipy.stats import chi2_contingency, mannwhitneyu
from IPython.display import display
import seaborn as sns
import matplotlib.pyplot as plt
import pingouin as pg
import statsmodels.api as sm

data = pd.read_csv('data/02-processed/nhis_2024_sample_adult_cleaned.csv')
df = data
data.head()

# ── Helper functions ──────────────────────────────────────────────────────────

def chi_square(predictor, outcome, d):
    contingency = pd.crosstab(d[predictor], d[outcome])
    chi2, p_value, dof, expected = stats.chi2_contingency(contingency)
    n = contingency.to_numpy().sum()
    w = np.sqrt(chi2 / n)
    print(predictor, '---', outcome)
    display(contingency)
    print('Chi square statistics: ', chi2)
    print('p_value: ', p_value)
    print('degree of freedom: ', dof)
    print('Cohen W: ', w)
    print('\n')

def chi_sq(data, pred, out):
    print(f"--- Testing Association with: {out} ---")
    contingency_table = pd.crosstab(data[pred], data[out])
    chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)
    print(f"Chi-Square Statistic: {chi2:.4f}")
    print(f"p-value: {p_val:.4e}")
    if p_val < 0.05:
        print("Result: Statistically Significant (p < 0.05). The distributions differ between groups.\n")
    else:
        print("Result: Not Statistically Significant. No strong evidence of a difference.\n")
    n = contingency_table.sum().sum()
    min_dim = min(contingency_table.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * min_dim))
    print(f"\nEffect Size:")
    print(f"  Cramér's V: {cramers_v:.4f}")
    print(f"\nCramér's V interpretation (Cohen's benchmarks):")
    if   cramers_v < 0.10: label = "negligible"
    elif cramers_v < 0.20: label = "small"
    elif cramers_v < 0.40: label = "medium"
    else:                  label = "large"
    print(f"  → {label} effect")

def mann_whitney(predictor, group1_name, group2_name, outcome, d):
    group1 = d[d[predictor] == group1_name][outcome]
    group2 = d[d[predictor] == group2_name][outcome]
    print(predictor, '---', outcome)
    display(pg.mwu(group1, group2))

def mwu(data, pred, out):
    group_yes = df[df[pred] == 1]
    group_no  = df[df[pred] == 0]
    print(f"--- Testing Differences in: {out} ---")
    u_stat, p_val = stats.mannwhitneyu(group_yes[out], group_no[out], alternative='two-sided')
    print(f"U Statistic: {u_stat:.4e}")
    print(f"p-value: {p_val:.4e}")
    if p_val < 0.05:
        print(f"Result: Statistically Significant (p < 0.05). The rank distributions of {out} differ.\n")
    else:
        print("Result: Not Statistically Significant. No strong evidence of a shift.\n")
    n1, n2 = len(group_yes[out]), len(group_no[out])
    r = 1 - (2 * u_stat) / (n1 * n2)
    print(f"Group '1': n={n1}, median={group_yes[out].median():.1f}, mean={group_yes[out].mean():.2f}")
    print(f"Group '0': n={n2}, median={group_no[out].median():.1f}, mean={group_no[out].mean():.2f}")
    print(f"\nEffect Size:")
    print(f"  Rank-biserial r: {r:.4e}")
    abs_r = abs(r)
    if   abs_r < 0.10: label = "negligible"
    elif abs_r < 0.30: label = "small"
    elif abs_r < 0.50: label = "medium"
    else:              label = "large"
    print(f"  → {label} effect")
    print(f"\nInterpretation: r = {r:.3f} means the '1' group tends to")
    print(f"  rank {'higher' if r > 0 else 'lower'} on {out} than '0'")

def cohen_w(chi2, n):
    return np.sqrt(chi2 / n)

def effect_size_mwu(u, n1, n2):
    rbc  = 1 - (2 * u) / (n1 * n2)
    cles = u / (n1 * n2)
    return round(rbc, 6), round(cles, 6)


# ── Section 1: housing_tenure vs outcomes ────────────────────────────────────

data_drop_other = data[data['housing_tenure'] != 'Other arrangement']

prop_df = data_drop_other.groupby('housing_tenure').usual_place_care.value_counts(normalize=True).rename('proportion').reset_index()
tenure_usual_plot = sns.barplot(data=prop_df, x='housing_tenure', y='proportion', hue='usual_place_care')
tenure_usual_plot.legend(title='usual_place_care', bbox_to_anchor=(1.02, 1))

prop_df = data_drop_other.groupby('housing_tenure').delayed_medical_care.value_counts(normalize=True).rename('proportion').reset_index()
prop_df['delayed_medical_care'] = prop_df['delayed_medical_care'].map({0: 'No', 1: 'Yes'})
tenure_delayed_plot = sns.barplot(data=prop_df, x='housing_tenure', y='proportion', hue='delayed_medical_care')
tenure_delayed_plot.legend(title='delayed_medical_care', bbox_to_anchor=(1.02, 1))

tenure_er_stacked = pd.crosstab(data_drop_other['housing_tenure'], data_drop_other['any_er_visit'], normalize='index')
ax = tenure_er_stacked.plot(kind='bar', stacked=True)
ax.legend(title='number_er_visit', bbox_to_anchor=(1.02, 1))

tenure_gap_stacked = pd.crosstab(data_drop_other['housing_tenure'], data_drop_other['doctor_gap_yr'], normalize='index')
ax = tenure_gap_stacked.plot(kind='bar', stacked=True)
ax.legend(title='doctor_gap_year', bbox_to_anchor=(1.02, 1))

tenure_health_stacked = pd.crosstab(data_drop_other['housing_tenure'], data_drop_other['general_health'], normalize='index')
ax = tenure_health_stacked.plot(kind='bar', stacked=True)
ax.legend(title='general_health', bbox_to_anchor=(1.02, 1))

chi_square('housing_tenure', 'usual_place_care', data_drop_other)
chi_square('housing_tenure', 'delayed_medical_care', data_drop_other)
mann_whitney('housing_tenure', 'Owned or being bought', 'Rented', 'general_health', data_drop_other)
mann_whitney('housing_tenure', 'Owned or being bought', 'Rented', 'any_er_visit', data_drop_other)
mann_whitney('housing_tenure', 'Owned or being bought', 'Rented', 'doctor_gap_yr', data_drop_other)


# ── Section 2: transportation_barrier vs outcomes ─────────────────────────────

fig, axes = plt.subplots(5, 1, figsize=(5, 25))
fig.suptitle('Transportation Barrier vs Healthcare Outcomes', fontsize=14)
plt.subplots_adjust(hspace=0.4)

plot_df = clean_nhis.copy()
plot_df['transport_label'] = plot_df['transportation_barrier'].map({0: 'No Barrier', 1: 'Has Barrier'})

# Plot 1
prop1 = plot_df.groupby('transport_label')['delayed_medical_care'].mean().reset_index()
prop1.columns = ['transport_label', 'proportion']
sns.barplot(data=prop1, x='transport_label', y='proportion', ax=axes[0])
axes[0].set_title('Delayed Medical Care')
axes[0].set_ylabel('Proportion with Delay')

# Plot 2
prop2 = plot_df.groupby(['transport_label', 'usual_place_care']).size().reset_index(name='count')
prop2['proportion'] = prop2.groupby('transport_label')['count'].transform(lambda x: x / x.sum())
sns.barplot(data=prop2, x='transport_label', y='proportion', hue='usual_place_care', ax=axes[1])

# Plot 3
plot_df['er_capped'] = plot_df['any_er_visit'].clip(upper=5)
sns.violinplot(data=plot_df, x='transport_label', y='er_capped', ax=axes[2])
axes[2].set_title('ER Visits')
axes[2].set_ylabel('Number of ER Visits')

# Plot 4
sns.violinplot(data=plot_df, x='transport_label', y='doctor_gap_yr', ax=axes[3])
axes[3].set_title('Doctor Gap')
axes[3].set_ylabel('Years Since Last Visit')

# Plot 5
health_labels = {1: 'Poor', 2: 'Fair', 3: 'Good', 4: 'Very Good', 5: 'Excellent'}
health_order  = ['Poor', 'Fair', 'Good', 'Very Good', 'Excellent']
plot_df['health_label'] = plot_df['general_health'].map(health_labels)
prop5 = plot_df.groupby(['transport_label', 'health_label']).size().reset_index(name='count')
prop5['proportion'] = prop5.groupby('transport_label')['count'].transform(lambda x: x / x.sum())
pivot5 = prop5.pivot(index='transport_label', columns='health_label', values='proportion')[health_order]
pivot5.plot(kind='bar', stacked=True, ax=axes[4])
axes[4].set_title('General Health Status')
axes[4].set_ylabel('Proportion')
axes[4].set_xticklabels(['No Barrier', 'Has Barrier'], rotation=0)
axes[4].legend(title='Health Status', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)
plt.show()

group0 = clean_nhis[clean_nhis['transportation_barrier'] == 0]
group1 = clean_nhis[clean_nhis['transportation_barrier'] == 1]

ct1 = pd.crosstab(clean_nhis['transportation_barrier'], clean_nhis['delayed_medical_care'])
display(ct1)
chi2_1, p1, dof1, _ = chi2_contingency(ct1)
n = ct1.values.sum()
display(pd.DataFrame({
    'Chi2':    [round(chi2_1, 4)],
    'p_value': [f'{p1:.4e}'],
    'df':      [dof1],
    'Cohen W': [round(cohen_w(chi2_1, n), 4)]
}))
print()

ct2 = pd.crosstab(clean_nhis['transportation_barrier'], clean_nhis['usual_place_care'])
display(ct2)
chi2_2, p2, dof2, _ = chi2_contingency(ct2)
n = ct2.values.sum()
display(pd.DataFrame({
    'Chi2':    [round(chi2_2, 4)],
    'p_value': [f'{p2:.4e}'],
    'df':      [dof2],
    'Cohen W': [round(cohen_w(chi2_2, n), 4)]
}))
print()

u3, p3 = mannwhitneyu(group0['general_health'].dropna(), group1['general_health'].dropna(), alternative='two-sided')
n0, n1 = len(group0['general_health'].dropna()), len(group1['general_health'].dropna())
rbc3, cles3 = effect_size_mwu(u3, n0, n1)
display(pd.DataFrame({
    'U_val':       [round(u3, 1)],
    'alternative': ['two-sided'],
    'p_val':       [f'{p3:.4e}'],
    'RBC':         [rbc3],
    'CLES':        [cles3]
}, index=['MWU']))
print()

u4, p4 = mannwhitneyu(group0['any_er_visit'].dropna(), group1['any_er_visit'].dropna(), alternative='two-sided')
n0, n1 = len(group0['any_er_visit'].dropna()), len(group1['any_er_visit'].dropna())
rbc4, cles4 = effect_size_mwu(u4, n0, n1)
display(pd.DataFrame({
    'U_val':       [round(u4, 1)],
    'alternative': ['two-sided'],
    'p_val':       [f'{p4:.4e}'],
    'RBC':         [rbc4],
    'CLES':        [cles4]
}, index=['MWU']))
print()

u5, p5 = mannwhitneyu(group0['doctor_gap_yr'].dropna(), group1['doctor_gap_yr'].dropna(), alternative='two-sided')
n0, n1 = len(group0['doctor_gap_yr'].dropna()), len(group1['doctor_gap_yr'].dropna())
rbc5, cles5 = effect_size_mwu(u5, n0, n1)
display(pd.DataFrame({
    'U_val':       [round(u5, 1)],
    'alternative': ['two-sided'],
    'p_val':       [f'{p5:.4e}'],
    'RBC':         [rbc5],
    'CLES':        [cles5]
}, index=['MWU']))


# ── Section 3: housing_cost_trouble vs outcomes ───────────────────────────────

print(pd.crosstab(df['housing_cost_trouble'], df['delayed_medical_care'], normalize='index') * 100)

sns.set_theme(style="whitegrid")
g = sns.catplot(
    data=df,
    x="housing_cost_trouble",
    hue="delayed_medical_care",
    kind="count",
    palette="muted",
    height=5,
    aspect=1.2
)
g.set_axis_labels("Housing Cost Trouble Status", "Count of Respondents")
g._legend.set_title("Delayed Medical Care")
plt.show()

chi_sq(df, "housing_cost_trouble", "delayed_medical_care")

cross_tab  = pd.crosstab(df['usual_place_care'], df['housing_cost_trouble'], normalize='columns') * 100
print(cross_tab)
plot_data  = cross_tab.T
ax = plot_data.plot(kind='bar', stacked=True, figsize=(9, 6), width=0.5, colormap='tab10')
plt.title("Usual Place of Care Distribution by Housing Trouble Status", fontsize=14, pad=15)
plt.xlabel("Housing Cost Trouble Status", fontsize=12)
plt.ylabel("Percentage within Group (%)", fontsize=12)
plt.xticks(rotation=0)
plt.legend(title="Usual Place of Care", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

chi_sq(df, "housing_cost_trouble", 'usual_place_care')

er_summary = df.groupby('housing_cost_trouble')['any_er_visit'].agg(
    Mean='mean', Median='median', Variance='var', Max='max', Total_n='count'
).reset_index()
print("--- ER Visit Summary Statistics ---")
print(er_summary.to_string(index=False))

sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 6))
sns.violinplot(
    data=df,
    x="housing_cost_trouble",
    hue="housing_cost_trouble",
    y="any_er_visit",
    palette="muted",
    inner="quart",
    cut=0
)
plt.title("Distribution of ER Visits by Housing Trouble Status", fontsize=14, pad=15)
plt.xlabel("Housing Cost Trouble Status", fontsize=12)
plt.ylabel("Number of ER Visits", fontsize=12)
plt.tight_layout()
plt.show()

chi_sq(df, "housing_cost_trouble", "any_er_visit")

er_summary = df.groupby('housing_cost_trouble')['doctor_gap_yr'].agg(
    Mean='mean', Median='median', Variance='var', Max='max', Total_n='count'
).reset_index()
print("--- ER Visit Summary Statistics ---")
print(er_summary.to_string(index=False))

sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 6))
sns.violinplot(
    data=df,
    x="housing_cost_trouble",
    hue="housing_cost_trouble",
    y="doctor_gap_yr",
    palette="muted",
    inner="quart",
    cut=0
)
plt.title("Distribution of doctor_gap_yr by Housing Trouble Status", fontsize=14, pad=15)
plt.xlabel("Housing Cost Trouble Status", fontsize=12)
plt.ylabel("doctor_gap_yr\t", fontsize=12)
plt.tight_layout()
plt.show()

chi_sq(df, "housing_cost_trouble", "doctor_gap_yr")

health_pct = pd.crosstab(df['housing_cost_trouble'], df['general_health'], normalize='index') * 100
print("--- General Health Proportions (%) by Housing Status ---")
print(health_pct.round(2))
print("\n--- Median Health Score (1-5 Scale) ---")
medians = df.groupby('housing_cost_trouble')['general_health'].median()
for group, median_val in medians.items():
    print(f"Housing Trouble [{group}]: Median Score = {median_val}")

labels_dict = {1: '1 (Poor)', 2: '2 (Fair)', 3: '3 (Good)', 4: '4 (Very Good)', 5: '5 (Excellent)'}
plot_data = health_pct.rename(columns=labels_dict)
ax = plot_data.plot(kind='barh', stacked=True, figsize=(10, 5), width=0.5, colormap='viridis_r')
plt.title("General Health Score Distribution by Housing Trouble Status", fontsize=14, pad=15)
plt.xlabel("Percentage within Housing Group (%)", fontsize=12)
plt.ylabel("Housing Cost Trouble Status", fontsize=12)
plt.xlim(0, 100)
plt.legend(title="General Health Score", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

mwu(df, "housing_cost_trouble", "general_health")
```
