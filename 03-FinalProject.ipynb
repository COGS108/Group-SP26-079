## Research Question

How do respondents’ delayed care due to unreliable transportation (boolean variable from survey) and housing vulnerability (Had trouble paying for housing? Residence owned or rented? Government paying part of rent?) affect their pattern of access healthcare (number of delayed medical care, number of ER visit, time since last saw doctor, general health status) in Delaware in 2024?

## Background and Prior Work

According to the World Health Organization (WHO), Social Determinants of Health (SDOH), which refers to non-medical factors that influence health outcomes, include conditions such as socioeconomic status, education, neighborhood and geographic location, employment, and broader structural forces such as social policies and systemic inequality <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). In a large study of a hospital-based financial assistance program at Kaiser Permanente, researchers found that providing financial assistance significantly increased healthcare utilization, including outpatient visits, emergency department encounters, and inpatient care <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3).

Within the system of SDOH, transportation and housing barriers are widely recognized as major obstacles to receiving healthcare, particularly among low-income populations and individuals with limited insurance coverage <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). Sixty-one studies across many states in the United States have shown that individuals without reliable transportation are more likely to miss appointments, delay care, or fail to receive recommended treatments <a name="cite_ref-4b"></a>[<sup>4</sup>](#cite_note-4). Similarly, a study of Medicaid patients in West Philadelphia, which is close to Delware, found that transportation difficulties were associated with missed primary care visits, particularly among low-income populations <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5). Through meta-analysis in the United States, Reid et al. found that individuals experiencing unstable housing are more likely to delay medical care, postpone medications, and experience difficulties accessing routine healthcare services. Housing instability has also been associated with fragmented healthcare utilization patterns and increased hospitalization rates <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6). These findings suggest that structural barriers, including both transportation and housing constraints, play a major role in shaping healthcare access and utilization patterns. So we choose Delware, where contains both urban and rural communities, as an area as interest to examine these structural barriers' effect.

Nonetheless, a systematic review of non-emergency medical transportation interventions found that although some programs reduced missed appointments, evidence regarding healthcare utilization, costs, and long-term health outcomes remained limited and inconclusive <a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7). Therefore, this project aims to examine and confirm how transportation-related barriers and housing vulnerability are associated with healthcare utilization patterns, highlighting the role of physical access in shaping general health outcomes.

## Hypothesis


The transportation barrier and housing vulnerability will affect the healthcare outcome. Populations with impeded transportation and uninsured housing may demonstrate a less frequent and standardized routine of accepting healthcare from hospitals and a worse general health status. High housing vulnerability may contribute more than high transportation barriers to the lower than average healthcare outcome.

### Count Model: Healthcare Access Disruption Score

To see the code responsible for Data analyses after EDA analysis see `Data Analysis -2nd part.ipynb`.

We constructed an `access_disruption_score` from five indicators: delayed medical care, not having a usual place of care, at least one ER visit, more than one year since last doctor visit, and poor/fair self-rated health. 
The access disruption score is a composite count variable summing five binary indicators (0 or 1 each), yielding a possible range of 0–5:

| Component | Coded as 1 if… |
|---|---|
| Delayed medical care | Reported delaying care |
| Usual place of care | No single or no usual place of care |
| ER visit | One or more ER visits in the past year |
| Doctor gap | More than one year since last doctor visit |
| General health | Fair or poor self-rated health |

A higher score indicates greater disruption to healthcare access.
The score ranges from 0 to 5 (only 3 individuals in the sample get a score of 5), where higher values indicate more healthcare access disruption. 


```python
dist = pd.read_csv('results/access_disruption_score_distribution.csv')
print('Access Disruption Score Distribution and Summary \n')
print(dist.round(4))
summ=pd.read_csv('results/access_disruption_score_summary.csv')
print('\n',summ.round(4))
```

We first use Logistic Regression to build classification model, but the unbalanced class lead to low accuracy, even with stratifying and survey weights, according to the confusion matrix ('results/classification_confusion_matrix.png') and classification report ('results/access_disruption_score_summary.csv'). For this reason, a preliminary generalized linear model (GLM) using Poisson regression with robust standard errors was fitted to assess the overall direction of key predictors. This keeps the model interpretable while avoiding the class-imbalance problem from the binary logistic regression. 

```python
full_key_results = pd.read_csv('results/access_disruption_poisson_key_results.csv', index_col=False)
full_key_results.at[0,'term'] = 'Other arrangement'
full_key_results.at[1,'term'] = 'Rented'
full_key_results.round(4)
```

The Poisson model gives a more detailed result and trend compared to exploratory analysis. Incident Rate Ratios (IRR) reflect the relative rate of a count outcome, with above 1 indicate increased rate and below 1 indicate decreased rate relative to the reference group. After adjusting for controls (age, poverty ratio, sex, race/ethnicity, education, and insurance), respondents with a transportation barrier had about **34% higher expected healthcare access disruption** than respondents without that barrier (IRR = 1.34, 95% CI [1.28, 1.40], p < 0.001). Respondents who had trouble paying housing costs had about **39% higher expected disruption** (IRR = 1.39, 95% CI [1.33, 1.45], p < 0.001). Compared with homeowners, renters had about **25% higher expected disruption** (IRR = 1.25, 95% CI [1.21, 1.29], p < 0.001).

Since the government rent assistance predictor only applied to renters, we stratified the sample to generate the renter only dataset to analyze. As a renter-only reliability check, government rent assistance was associated with a small increase in expected disruption, but this estimate was only borderline significant (IRR = 1.06, 95% CI [1.00, 1.12], p = 0.056). This suggests rent assistance may identify a more vulnerable subgroup of renters, but it is not as strong or stable a predictor as transportation barriers and housing cost trouble. Overall, this model appears more appropriate than a classification model because it preserves the direction and magnitude of the housing and transportation effects without being dominated by the majority class.

### Additional Analyses Addressing Specific Aspects of the Outcomes

The main count model summarizes overall healthcare access disruption, but our research question includes several distinct aspects of healthcare access and health status. To avoid relying on a single composite outcome, we use Poisson GLM for count outcomes (doctor_gap_yr, any_er_visit, general_health) and binomial GLM for binary or categorical outcomes (delayed_medical_care, usual_palace_care) to inspect each outcome's regression model with predictors.

```python
aspect=pd.read_csv('results/aspect_key_results.csv').round(4)
term_map = {
    'C(housing_tenure, Treatment(reference="Owned or being bought"))[T.Rented]': 'Renter',
    'C(housing_tenure, Treatment(reference="Owned or being bought"))[T.Other arrangement]': 'Other arrangement'
}
aspect['term'] = aspect['term'].replace(term_map)
aspect
```

Odds Ratios (OR) reflect the relative odds of a binary outcome occurring, where odds represent the ratio of the probability of an event happening to it not happening, with above 1 indicate increased odds and below 1 indicate decreased odds relative to the reference group. Across all five outcomes, housing cost trouble emerged as the stronger and more consistent predictor than transportation barriers: it nearly triples the odds of delayed care (OR 3.28) and substantially raises ER visit rates (IRR 1.66), while transportation barriers show meaningful but smaller effects in the same direction. Renter status is associated with worse outcomes across the board compared to owners — lower odds of having a usual place of care (OR 0.63) and higher ER utilization (IRR 1.26) — though effects on doctor gap years and general health are modest. Doctor gap year showed generally small, mostly non-significant associations with housing and transportation predictors, suggesting these barriers affect care access and utilization more than appointment frequency. Importantly, each predictor — transportation barrier, housing cost trouble, housing tenure, and government rent assistance — significantly affects at least one outcome, confirming that all of them meaningfully contribute to the variation in the overall access disruption score at this stage of analysis.
The mixed pattern of rent assistance observed in the individual outcome models — protective for care access outcomes but associated with higher ER utilization and worse health scores — is consistent with the main model finding that rent assistance identifies a more vulnerable renter subgroup rather than acting as a strong independent predictor, supporting the interpretation that recipients tend to have greater underlying need.

### Examine the dataset with most dominant ethnicity (White non-Hispanic) removed

The white non-Hispanic ethnicity compose the majority of the sample. To examine whether the patterns between predictors and outcomes in other ethnicities are hidden by the pattern in white non-Hispanic population, and whether there's difference in housing and transportation's effect between the major population in Delware (white non-Hispanic) and the minor populations, we substract white non-Hispanic individuals from datasets to perform Poisson regression.  

```python
nonwhite_key_results = pd.read_csv('results/access_non-white_subsample_key_poisson_results.csv', index_col=False)
nonwhite_key_results.at[0,'term'] = 'Other arrangement'
nonwhite_key_results.at[1,'term'] = 'Rented'
nonwhite_key_results.round(4)
```

After removing the `white non-Hispanic` from the samples, the model still predicts that populations without ownership of house, have transportation barrier, and have housing cost trouble will increase the disruption score, according to larger than 1 IRR value and less than 0.05 p_value for most predictors. Therefore, a similar pattern between socioeconomic predictors and access disruption score in the survey both exists for full population and population that only contain minor ethnicity. However, the IRR score for each predictor is generally lower than those from models of the full sample, indicating that the race ethnicity will affect how housing and transportation impact people's access of healthcare. Specifically, transportation and housing condition may have less effect on non-white population than white-population, but further analysis is needed to make the conclusion and prove that the difference is not due to chance.

### Prediction: Expected Healthcare Access Disruption

According to previous analysis, each predictor significantly affects at least 1 outcomes, so all of them contribute to the variation of the access disruption score at this stage of analysis. Also, the pattern we observed in the the model exists for both the dominant ethnicity and minors. Therefore, the Poisson GLM on the full dataset is relatively fair and can be potentially used for prediction. We used Poisson GLM to predict access disruption score for scenarios. We compared the Poisson GLM against a Train Mean baseline model that predicts the average Access Disruption Score for all respondents. Unlike the baseline, the Poisson GLM incorporates predictors to generate individualized predictions. The comparison allows us to evaluate whether these social determinants provide meaningful information beyond the overall population average.

```python
predicted_results = pd.read_csv('results/access_disruption_prediction_performance.csv', index_col=False)
scenario_prediction = pd.read_csv('results/access_disruption_scenario_predictions.csv', index_col=False)
print(predicted_results.round(4))
scenario_prediction.round(4)
```

The prediction results show that the Poisson GLM performs slightly better than a simple baseline that predicts the training-set mean for every person. On the test set, the model reduced MAE from 0.708 to 0.650 and RMSE from 0.834 to 0.777. Due to the modest improvement, we could conclude that housing and transportation factors contribute to explaining variation in healthcare disruption but should not frame the model as a strong individual-level prediction tool.

The scenario predictions are more useful for interpretation. For a typical insured respondent with median age and poverty ratio, the expected access disruption score is about **0.51** when there is no transportation or housing barrier. With a transportation barrier only, the expected score rises to **0.68**. With housing cost trouble only, it rises to **0.71**. With both barriers, it rises to **0.95**. If the person is also a renter, the expected score rises further to about **1.19**. In plain terms, the model predicts that overlapping transportation and housing barriers create substantially higher expected healthcare access disruption than either barrier alone.

## Discusison and Conclusion

### Conclusion

This study use basic exploratory analysis (Chi-square, Mann Whitney U test) and inferential and predictive model (GLM) to approach the 2024 NHIS survey, and the results from models generally accord wit the exploratory analysis. The study found that housing and transportation barriers are consistently associated with reduced healthcare access and poorer health outcomes, and they also independently contribute to health access outcomes. Across multiple healthcare access measures, individuals experiencing housing cost burden, transportation barriers, or lacking home ownership were more likely to experience disruptions in care. Poisson Regression analysis further showed that transportation barriers, housing cost burden, and non-homeownership were associated with approximately 30% higher Access Disruption Scores, while government rent assistance contribute to a small increase in the scores. 

Overall, these findings reinforce the importance of Social Determinants of Health in shaping healthcare utilization. The results suggest that improving housing stability and reducing transportation barriers may be important strategies for increasing healthcare access and reducing health disparities. Moreover, the greater disruption observed when multiple barriers co-occur highlights the need for interventions that address socioeconomic challenges in combination rather than in isolation.

### Discussion

This study extends prior research on Social Determinants of Health (SDOH) by examining how housing and transportation barriers are associated with healthcare access outcomes in Delaware. Previous studies have shown that transportation difficulties contribute to missed appointments and delayed care, while housing instability is linked to fragmented healthcare utilization and poorer health outcomes. Our findings support these conclusions and provide quantitative estimates of their relative impact using NHIS data.

The exploratory analyses demonstrated that housing tenure, transportation barriers, housing cost burden, and government housing assistance were all significantly associated with healthcare access outcomes, including years since last doctor visit, emergency room utilization, delayed medical care, usual place of care, and self-reported health status. Although many individual associations had small effect sizes, the consistency of the relationships across multiple outcomes suggests that these socioeconomic factors play an important role in shaping healthcare utilization patterns.

To summarize overall healthcare accessibility, we constructed an Access Disruption Score and modeled it using Poisson Regression. Transportation barriers, housing cost burden, and non-homeownership were each associated with approximately 30% higher expected disruption scores. These findings are consistent with the literature reviewed in the Background section. Syed et al. (2013) reported that transportation barriers consistently reduce healthcare utilization, while Reid et al. (2008) found that housing instability is associated with fragmented patterns of care. Our results provide additional evidence supporting these conclusions within a Delaware population. Unexpectedly, we found government housing assistance was associated with a small increase in the Access Disruption Score, which is likely reflects the greater underlying socioeconomic challenges faced by assistance recipients rather than a negative effect of the program itself. Further longitudinal research is needed to evaluate the causal impact of housing assistance on healthcare access. 

One particularly notable finding through Poisson GLM was that the combined presence of housing cost burden and transportation barriers was associated with greater healthcare disruption than either factor alone. This suggests that multiple socioeconomic disadvantages may interact or accumulate, creating additional obstacles to healthcare access. Rather than acting independently, housing and transportation conditions may jointly influence an individual's ability to obtain timely and consistent medical care.

The analysis of the Non-White sample subset produced results that were broadly consistent with those observed in the full sample, indicating that housing and transportation barriers remain important predictors of healthcare access across racial and ethnic groups. However, the estimated effects were somewhat smaller among Non-White respondents. Several explanations are possible. The smaller subgroup sample size may have reduced statistical power, making effect estimates less stable. Alternatively, differences in social support systems, healthcare-seeking behaviors, or community resources may partially buffer the impact of housing and transportation barriers. Because formal interaction analyses were not conducted, these explanations remain speculative and warrant further investigation.

Several limitations should be considered when interpreting these findings. First, the dataset has classes that are highly imbalanced due to the non even distribution of people with different conditions in the population. Second, NHIS data are cross-sectional, preventing conclusions about causality. Poor health may contribute to housing instability or transportation difficulties just as easily as these barriers may contribute to poor healthcare access. Third, all measures were self-reported and therefore subject to recall and reporting bias. Fourth, survey-based sampling may underrepresent highly vulnerable populations, including unhoused individuals, people without reliable phone access, and institutionalized populations. As these groups are likely to experience the greatest transportation and housing challenges, the true effects may be larger than those estimated in this study. Finally, the public-use NHIS dataset suppresses detailed geographic information, limiting the ability to investigate neighborhood-level or community-level variation.

Future research should focus on linking survey data with administrative healthcare records to validate self-reported outcomes and improve measurement accuracy. Longitudinal studies would also be valuable because they could determine whether changes in housing or transportation conditions precede changes in healthcare utilization, helping establish causal relationships that cannot be identified using cross-sectional data alone.
