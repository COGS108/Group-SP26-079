### Count Model: Healthcare Access Disruption Score

Because the binary delayed-care outcome is highly imbalanced, we also modeled a broader count-style outcome instead of treating the task as a pure classification problem. 

We first considered a negative binomial model, but the fitted dispersion parameter was effectively zero, meaning the score was not meaningfully overdispersed. For this reason, the more stable pilot model reported below is a Poisson regression with robust standard errors. This keeps the model interpretable while avoiding the class-imbalance problem from the binary logistic regression. To represent patient's access of healthcare, we also used the access_disruption_score to combine the 5 outcomes.


```python
full_key_results = pd.read_csv('results/access_disruption_poisson_key_results.csv', index_col=False)
full_key_results.at[0,'term'] = 'Other arrangement'
full_key_results.at[1,'term'] = 'Rented'
full_key_results.round(4)
```

The pilot count model gives a more detailed result and trend compared to exploratory analysis. After adjusting for controls (age, poverty ratio, sex, race/ethnicity, education, and insurance), respondents with a transportation barrier had about **34% higher expected healthcare access disruption** than respondents without that barrier (IRR = 1.34, 95% CI [1.28, 1.40], p < 0.001). Respondents who had trouble paying housing costs had about **39% higher expected disruption** (IRR = 1.39, 95% CI [1.33, 1.45], p < 0.001). Compared with homeowners, renters had about **25% higher expected disruption** (IRR = 1.25, 95% CI [1.21, 1.29], p < 0.001).

Since the government rent assistance predictor only applied to renters, we stratified the sample to generate the renter only dataset to analyze. As a renter-only reliability check, government rent assistance was associated with a small increase in expected disruption, but this estimate was only borderline significant (IRR = 1.06, 95% CI [1.00, 1.12], p = 0.056). This suggests rent assistance may identify a more vulnerable subgroup of renters, but it is not as strong or stable a predictor as transportation barriers and housing cost trouble. Overall, this model appears more appropriate than a classification model because it preserves the direction and magnitude of the housing and transportation effects without being dominated by the majority class.

This study analyzed 2024 NHIS health outcome data and examined how housing- and transportation-related social determinants of health influence healthcare access. The analysis focused on four predictors: housing tenure, transportation barriers, housing cost burden, and government housing assistance. Healthcare access was evaluated using five outcomes: years since last doctor visit, number of emergency room (ER) visits, delayed medical care, usual place of care, and self-reported general health status.

Exploratory analyses showed that all four predictors were significantly associated with healthcare access outcomes. Individuals experiencing transportation barriers, housing cost difficulties, or lacking home ownership generally reported poorer healthcare access and worse health outcomes. Although many effect sizes were statistically small, the large sample size provides strong evidence of a consistent relationship between socioeconomic conditions and healthcare utilization.

To better capture overall healthcare accessibility, an Access Disruption Score was created by combining the five healthcare-related outcomes. A Poisson Regression model was then used to estimate the expected level of healthcare disruption. The results indicated that transportation barriers, housing cost burden, and non-homeownership were associated with approximately 30% higher disruption scores compared with baseline groups. However, receiving government housing assistance was associated with only a modest increase of approximately 5%. The model also suggests that the combined presence of housing cost burden and transportation barriers is associated with even greater healthcare disruption than either factor alone. This finding highlights the possibility that multiple socioeconomic disadvantages may exert an additive or compounding effect on healthcare access. To investigate whether these patterns differed across racial and ethnic groups, the same Poisson Regression analysis was performed on a Non-White sample, excluding White non-Hispanic respondents. The overall pattern remained consistent with the full-sample analysis, indicating that housing and transportation factors continued to be important predictors of healthcare disruption. However, the estimated effects of all predictors were somewhat smaller in the Non-White sample. If future studies confirm that these differences are statistically significant, the findings may suggest that race and ethnicity play an important moderating role in the relationship between socioeconomic conditions and healthcare access. 

Overall, the findings demonstrate that individuals without home ownership, those experiencing housing cost difficulties, and those facing transportation barriers are more likely to experience disrupted and unstable access to healthcare, which may ultimately contribute to poorer health outcomes. The small positive association between government housing assistance and healthcare disruption should not be interpreted as a harmful effect of assistance itself. Instead, it likely reflects the fact that individuals receiving assistance are already among the most socioeconomically vulnerable populations. These findings underscore the importance of addressing housing and transportation barriers as part of broader efforts to improve healthcare accessibility and reduce health disparities.

The result we discovered align with prior literature. Syed et al. (2013) showed transportation barriers consistently lead to missed appointments and delayed care among studies, and our Poisson model puts a concrete number on that: a ~34% increase in disruption score, holding other factors constant. Reid et al. (2008) similarly linked housing instability to fragmented care patterns, consistent with our finding that housing cost trouble carried the largest single IRR (1.39) in the model. That two independent lines of evidence point in the same direction makes the associations harder to dismiss as dataset-specific noise.

The smaller effect sizes observed in non-white subsample worth particular attention. One explanation is that Non-White communities have long navigated structural disadvantages and may rely more heavily on informal support networks that partially diminish the need for individual transportation or housing barriers. The other explanation is that Non-White subsample is smaller and more diverse, which would reduce power to detect effects of the same magnitude as the full sample. Without formal interaction tests, it is difficult to suggest which one is the real reason.

Our study contains a few limitations. Since NHIS is a cross-sectional research, cannot rule out reverse causation, poorer health itself may contribute to housing instability rather than the other way around. Moreover, The survey is also self-reported and telephone-based, which likely underrepresents the most vulnerable groups: unhoused individuals, people without phone access, and institutionalized populations. These are possibly the people most affected by transportation and housing barriers, so our estimates are probably conservative. Geographic detail is also suppressed in the public-use file, preventing any finer-grained local analysis within Delaware.

The most useful next step would be pairing survey data like NHIS with administrative claims records to cross-validate self-reported outcomes. Longitudinal tracking would also help establish whether housing or transportation changes actually precede shifts in healthcare utilization, which is the causal question this study ultimately cannot answer.
